```python
new['Touchscreen'] = new['ScreenResolution'].apply(lambda x:1 if 'Touchscreen' in x else 0)
new['Ips'] = new['ScreenResolution'].apply(lambda x:1 if 'IPS' in x else 0)


'''
regular expression(regex)

r''          -> Raw string, used to write regular expressions. used in this case to tell python to run the backslash as ordinary backslash
()           -> Capturing group; extract() returns only the text inside these brackets.
\d+          -> Matches one or more digits (in this case., 1920 ).
x            -> Matches the literal character 'x' between the width and height.
\d+          -> Matches the next one or more digits after the 'x' (in this case., 1080 ).
Overall, this extracts screen resolutions in the format WidthxHeight (e.g., 1920x1080).
'''
new['Resolution'] = new['ScreenResolution'].str.extract(r'(\d+x\d+)')
```

```python
#to seperate dependent variable from independent variable

x= new.drop(columns = 'Price')
y= new['Price']
```

```python
#these are models and pipelines that I require in my analysis

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

#these are the 4 models I will be using to train and test my data
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

#to seperate the tet data from the training data
from sklearn.model_selection import train_test_split
x_train,x_test, y_train, y_test = train_test_split(x, y, test_size = 0.3, random_state = 42)

#Evaluation metrics
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
#i want to convert the categorical columns to numbers  what my machine would be able to read and understand, I start by creating a variable that holds the column that I want to transform

cat_cols = ['Company', 'TypeName', 'CPU_brand', 'Gpu_brand', 'OpSys',]
```

```python
#next i introduce my transformwr in a variable called transform
#the transformer take 3 arguments, one is the name of the transformation, the next is what transformation method or process i am deploying, the last is what I am tranforming
#remainder tells it to pass through  every other column

transform = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)
    ],
    remainder='passthrough'
)
```

```python
#i am creating a function that I can recall whenever I want to introduc a new model to my pipeline 

def create_pipeline(model):
    return Pipeline([
        ('preprocessor', transform),
        ('model', model)
    ])
```

```python
#now, i'm specifying the model I want to employ, and insertingit into a pipeline that I had already impported

#LinearRegression()
pipeline = create_pipeline(LinearRegression())
```

```python
#YOU ARE TELLING THE MACHINE TO APPLY THE MODEL IN YOUR PIPELINE TO STUDY THE TRAINING DATA (x AND y)

pipeline.fit(x_train, y_train)
y_pred = pipeline.predict(x_test)
y_pred[:5]
```

```python
#a table to compare actual with predicted 

comparison = pd.DataFrame({
    "Actual": y_test,
    "Predicted": y_pred
})

comparison.head(20)
```

```python
R2_score = r2_score(y_test, y_pred)
MAE = mean_absolute_error(y_test, y_pred)
RMSE = np.sqrt(mean_squared_error(y_test, y_pred))

print(f"Your R² Score : {R2_score}")
print(f"Your MAE: {MAE}")
print(f"Your RMSE: {RMSE}")
```
